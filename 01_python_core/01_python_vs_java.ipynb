```python
from dataclasses import dataclass

# Java: class Person { String name; int age; ... }
@dataclass
class Person:
    name: str
    age: int

people = [Person("Asha", 34), Person("Ben", 31), Person("Chloe", 41)]

# Java: people.stream().filter(p -> p.age > 30).map(p -> p.name).toList()
names_over_30 = [p.name for p in people if p.age > 30]

print(names_over_30)
print(f"{len(people)} people, average age {sum(p.age for p in people) / len(people):.1f}")
```
